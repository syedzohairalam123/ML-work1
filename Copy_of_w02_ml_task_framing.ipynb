## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*


Why ML Beats Hardcoded Heuristics:

Multi-Signal Interaction: A simple rule like "if CTR drops by 10%, flag it" misses complex patterns. Decay often involves non-linear interactions between query volatility, position drifts, and page age.

Client Variance: Different domains have different baseline CTRs and query distributions. Machine Learning models learn non-linear decision boundaries that generalize across diverse clients without manual threshold tuning.

Reduced False Alarms: Fixed rules generate high false positives during normal seasonal traffic fluctuations. ML incorporates historical trend signals to isolate genuine decay from temporary noise.
